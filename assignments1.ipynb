```python
def val_user(name):
    name = name.strip()
    
    if len(name) >= 5 and name.isalnum():
        return name.lower()
    else:
        return "Invalid"
name1 = " CodeMaster99 "
name2 = " Python_Dev "
print(val_user(name1))
print(val_user(name2))
```

```python
def decode(s):
    return " ".join(s.split('-')[::-1]).upper()
s = "charlie-bravo-alpha"
print(decode(s))
```

```python
def clean(s):
    return ", ".join(x.strip().title() for x in s.split(';'))

s = " john doe ; jAnE sMiTh; ALAN tUrInG "
print(clean(s))
```

```python
def fix(s):
    return " ".join(s.split()).capitalize()

s = "   python    is   a  very        powerful   language   "
print(fix(s))
```

```python
def check(s):
    c = "".join(x.lower() for x in s if x.isalnum())
    return c == c[::-1]

s = "Was it a car or a cat I saw?"
print(check(s))

s = "Python is fun"
print(check(s))
```

```python
def extract(s):
    return [x[1:].lower() for x in s.split() if x.startswith('#')]

s = "Loving the SUNSET! #Nature #Sunset_Vibes #beautiful"
print(extract(s))
```

```python
def find(s):
    return list(dict.fromkeys(x.split('.')[-1].strip().upper() for x in s.split(',')))

s = "report.pdf, image.JPG, data.csv, notes.txt, photo.jpg"
print(find(s))
```

```python
def get(s):
    if '@' not in s:
        return "Invalid"
    return s.split('@')[1].split('.')[0].title()

s = "developer_99@microsoft.com"
print(get(s))

s = "admin_at_website.org"
print(get(s))
```

```python
def check(a, b):
    return sorted(a.replace(" ", "").lower()) == sorted(b.replace(" ", "").lower())

a = "Dormitory"
b = "Dirty room"
print(check(a, b))
```

```python
def format(s):
    d = "".join(x for x in s if x.isdigit())
    if len(d) == 10:
        return f"({d[:3]}) {d[3:6]}-{d[6:]}"
    return "Invalid Number"

s = " +1 (555) 123.4567 "
print(format(s))

s = "555-987-6543"
print(format(s))
```

```python
def check(s):
    c = s.strip().upper()
    if len(c) == 8 and c[:3].isalpha() and c[3] == '-' and c[4:].isdigit():
        return f"Valid: {c}"
    return "Invalid SKU"

s = " xYz-9876 "
print(check(s))

s = "AB-12345"
print(check(s))

s = "A@C-1234"
print(check(s))
```
