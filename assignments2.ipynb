```python
def check(s):
    c = s.strip().title()
    if c.replace(" ", "").isalpha():
        return c
    return "Invalid city"

s = " nEw dELhi "
print(check(s))
```

```python
def calc(s):
    p = [int(x) for x in s.split(',')]
    return f"Total : {sum(p)}\nHighest : {max(p)}\nAverage : {sum(p)/len(p):.2f}"

s = " 120, 45 , 300 ,80, 95 "
print(calc(s))
```

```python
def validate(s):
    c = s.strip().lower()
    if len(c) >= 5 and len(c) <= 12 and c.isalnum() and c[0].isalpha():
        return "Valid username"
    return "Invalid username"

s = " Rahul_123 "
print(validate(s))

s = "rahul123"
print(validate(s))
```

```python
def stats(s):
    w = s.split()
    v = sum(1 for x in w if x[-1].lower() in 'aeiou')
    l = max(w, key=len)
    return f"Total words: {len(w)}\nEnds with vowel: {v}\nLongest word: {l}"

s = "Python is a powerful programming language"
print(stats(s))
```

```python
def count(s):
    d = {}
    for x in s.replace(" ", "").lower():
        d[x] = d.get(x, 0) + 1
    return d

s = "Aba c"
print(count(s))
```

```python
def invert(d):
    r = {}
    for k, v in d.items():
        r.setdefault(v, []).append(k)
    return r

sales = {"mon": 200, "tue": 350, "wed": 200}
print(invert(sales))
```

```python
def merge(a, b):
    r = dict(a)
    for k, v in b.items():
        r[k] = r.get(k, 0) + v
    return r

test1 = {"amit": 40, "bina": 35}
test2 = {"amit": 30, "chirag": 45}
print(merge(test1, test2))
```

```python
def show(e):
    for n, s in sorted(e, key=lambda x: x[1], reverse=True):
        print(f"{n} earns Rs. {s}")

employees = [("Ravi", 55000), ("Sita", 72000), ("Amit", 48000)]
show(employees)
```

```python
def parse(m):
    print("--- (a) Ratings > 8.1 ---")
    for x in m:
        if x["rating"] > 8.1:
            print(x["title"])
            
    print("\n--- (b) Genres ---")
    print(list(set(x["genre"] for x in m)))
    
    print("\n--- (c) Grouped by Rating ---")
    d = {}
    for x in m:
        d.setdefault(x["rating"], []).append(x["title"])
    for k, v in d.items():
        print(f"Rating {k}: {', '.join(v)}")

movies = [
    {"title": "Movie A", "rating": 8.2, "genre": "Thriller"},
    {"title": "Movie B", "rating": 8.4, "genre": "Sports"},
    {"title": "Movie C", "rating": 8.1, "genre": "Comedy"},
    {"title": "Movie D", "rating": 8.2, "genre": "Drama"}
]
parse(movies)
```
